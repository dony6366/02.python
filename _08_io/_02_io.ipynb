## 파일 쓰기

```python
f = open('helloworld2.txt','w',encoding='utf-8')
print(f, type(f))

text = '''
아
오늘 일이 안풀리네
삼재인가...
hoo...
'''
f.write(text) # 지정된 파일에  텍스트출력
# 지정된 파ㅣㄹ이 존재하지 않으면 파일 생성
# 기존 파일이 존재한다면 덮어쓰기

f.close() # 파일 연결에 사용된 메모리 반환
```

```python
# with절
# with 옆에 작성된 open() 구문에서 사용한 파일 연결을
# with절 종료 시 자동으로 close()
with open('helloworld2.txt','w',encoding='utf-8') as f:
    f.write('with절을 이용하면 자동 close() 된다')
```

## 파일 읽기

```python
with open('helloworld.txt','r',encoding='utf-8') as f:
    print(f.read()) # 전부 읽어와서 출력
```

```python
# 한 줄씩 읽기
with open('helloworld.txt','r',encoding='utf-8') as f:
    for i, line in enumerate(f):
        print(i+1, line, end='') # end = '' -> print()완료 후 마지막 문자 빈칸
```

```python
# 파일 이어쓰기('a' append 모드)
with open('helloworld.txt','a',encoding='utf-8') as f:
    f.write(''' 안녕하세요
저는 최경돈이라고 합니다.
제가 과연 이 수업을 잘 마칠 수 있을까요???😒😒😒''')

with open('helloworld.txt','r',encoding='utf-8') as f:
    print(f.read()) #이어쓰기 확인
```

```python
# 파일 복사
# 1. 파일 읽기
with open('helloworld.txt','r',encoding='utf-8') as f:
    text = f.read() # with 절에 손온헌 뵨슈눈 지역변수 X  -> 전역 사용 가능

# 2. 파일 쓰기
with open('helloworld_copy.txt','w',encoding='utf-8') as f:
    f.write(text)
    print('파일 복사 완료')
```

```python
# 이미지 파일 복사(아무 파일이나 가능)
# r, w, a == 문자 전용 읽기/쓰기/ 이어쓰기 모드 (2바이트씩 읽음)
# rn, wb == 바이트 단위 읽기/쓰기 모드
with open('hong.jpeg','rb') as f:
   # print(f.read())
    image_bytes = f.read()

with open('copy_hong.jpeg','wb') as f:
    f.write(image_bytes)
```

```python
# CSV 읽기
# import csv.DictReader(f)
#import csv.DictWtiter(f)
import csv
with open('users.csv','r',encoding='utf-8') as f:
    reader = csv.DictReader(f) # DictReader 객체 생성

    # users = []
    # for row in reader:
    #     print(row, type(row))
    #     users.append(row)

    users = [row for row in reader]

    print(users)
```

```python
#csv 쓰기
with open('users.csv','w',encoding='utf-8', newline='') as f:
    # csv 쓰기 객체 생성
    # - f : 쓸 파일
    # - fieldnames : 첫 줄에 작성될 헤더(컬럼명)
    writer = csv.DictWriter(f, fieldnames=['id','name','email'])
    writer.writeheader() # writeheader으로 헤더 작성

    users.append({'id':400, 'name':'이순신', 'email':'leess@gamil.com' })
    users.append({'id':500, 'name':'강감찬', 'email':'kkc@gamil.com' })

    #여러 줄 쓰기
    #writer.writerows(users)


    # 한 줄 씩 쓰기
    for user in users:
        writer.writerow(user)
```
